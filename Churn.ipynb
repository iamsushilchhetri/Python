```python
# Lets create subplot, done using chatgpt


# List of columns to plot
cols = [
    'PhoneService', 'MultipleLines', 'InternetService', 
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Set up a 3x3 grid
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(18, 15))

# Flatten axes array for easy iteration (since it's a 2D array [3,3])
axes = axes.flatten()

for i, column in enumerate(cols):
    # Plotting each column
    # hue='Churn' shows how each category relates to customer churn
    sns.countplot(data=df, x=column, hue='Churn', ax=axes[i], palette='viridis')
    
    axes[i].set_title(f'Count Plot of {column}')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Count')
    axes[i].legend(title='Churn', loc='upper right')

# Clean up layout to prevent overlapping
plt.tight_layout()
plt.show()
```

```python
plt.savefig('Subplots.png')
"""
Visualizing Categorical Features vs. Churn
This script generates a 3x3 grid of countplots to analyze how nine different 
categorical services (like Internet, Tech Support, etc.) impact customer churn. 
It loops through the features, automatically plotting the distribution of each 
and splitting the bars by 'Churn' status to easily spot high-risk categories.
"""
```
