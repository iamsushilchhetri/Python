```python
import numpy as np
import matplotlib.pyplot as plt

# Positive relationship
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 8, 10])

x_mean = np.mean(x)
y_mean = np.mean(y)

cov = np.sum((x - x_mean) * (y - y_mean)) / (len(x) - 1) # formula to find the covariance
print("Positive Covariance:", cov)

plt.scatter(x, y)

plt.axvline(x_mean, linestyle='--')
plt.axhline(y_mean, linestyle='--')

# rectangles
for i in range(len(x)):
    plt.plot([x[i], x[i]], [y_mean, y[i]])
    plt.plot([x_mean, x[i]], [y_mean, y_mean])

plt.title("Positive Covariance")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()

plt.show()
```
