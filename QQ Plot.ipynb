```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# set seed for reproducibility
np.random.seed(42)

# 1. Normal data
normal_data = np.random.normal(loc=0, scale=1, size=1000)

# 2. Too peaked (high kurtosis)
peaked_data = np.random.laplace(loc=0, scale=1, size=1000)

# 3. Skewed data (right skew)
skewed_data = np.random.exponential(scale=1, size=1000)

# Create subplots
fig, axes = plt.subplots(3, 2, figsize=(10, 12))

# ---- Normal ----
axes[0, 0].hist(normal_data, bins=30, density=True)
axes[0, 0].set_title("Normally Distributed Data")

stats.probplot(normal_data, dist="norm", plot=axes[0, 1])
axes[0, 1].set_title("Normal Q-Q Plot")

# ---- Peaked ----
axes[1, 0].hist(peaked_data, bins=30, density=True)
axes[1, 0].set_title("Data too peaked in middle")

stats.probplot(peaked_data, dist="norm", plot=axes[1, 1])
axes[1, 1].set_title("Normal Q-Q Plot")

# ---- Skewed ----
axes[2, 0].hist(skewed_data, bins=30, density=True)
axes[2, 0].set_title("Skewed Data")

stats.probplot(skewed_data, dist="norm", plot=axes[2, 1])
axes[2, 1].set_title("Normal Q-Q Plot")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

np.random.seed(42)

# datasets
normal = np.random.normal(0, 1, 1000)
right_skew = np.random.exponential(1, 1000)
left_skew = -np.random.exponential(1, 1000)
fat_tails = np.random.standard_t(df=2, size=1000)
thin_tails = np.random.uniform(-2, 2, 1000)

data_list = [
    ("Normal", normal),
    ("Right Skewed", right_skew),
    ("Left Skewed", left_skew),
    ("Fat Tails", fat_tails),
    ("Thin Tails", thin_tails)
]

fig, axes = plt.subplots(5, 2, figsize=(10, 18))

for i, (title, data) in enumerate(data_list):
    axes[i, 0].hist(data, bins=30, density=True)
    axes[i, 0].set_title(title)

    stats.probplot(data, dist="norm", plot=axes[i, 1])
    axes[i, 1].set_title(f"{title} Q-Q Plot")

plt.tight_layout()
plt.show()
```

```python
np.random.seed(42)

# shape parameter (alpha)
alpha = 2.5  

# generate data
pareto_data = np.random.pareto(alpha, 1000)

# shift to make it more realistic (optional)
pareto_data = pareto_data + 1

# Histogram
plt.hist(pareto_data, bins=50, density=True)
plt.title("Pareto Distribution (Histogram)")
plt.show()

# Q-Q Plot
stats.probplot(pareto_data, dist="norm", plot=plt)
plt.title("Pareto Q-Q Plot")
plt.show()
```
